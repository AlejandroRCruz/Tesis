```python
import pandas as pd

# 24 frames por segundo son 0.0416666 segundos
# en milisegundos son 41.6666 milisegundos, por frame.
# Archivos a comparar
archivos = [
    "resultados_pose.csv",
    "resultados_pose_1280.csv",
    "resultados_pose_img_resized.csv",
    "resultados_pose_raspberry_2_640.csv",
    "resultados_pose_raspberry.csv",
    "hombro_cadera_hailo.csv"
]

for archivo in archivos:
    try:
        df = pd.read_csv(archivo)

        # Convertir tiempos a float
        df["tiempo_ms"] = pd.to_numeric(df["tiempo_ms"], errors="coerce")

        # Agrupar por frame y tomar un solo tiempo por frame
        tiempos_por_frame = df.groupby("frame_id")["tiempo_ms"].first()

        # Calcular suma y promedio reales
        suma = tiempos_por_frame.sum()
        promedio = tiempos_por_frame.mean()
        fps_eff = 1000.0 / promedio if promedio > 0 else float("nan")

        print(f"{archivo}")
        print(f"   → Suma total de tiempos: {suma:.2f} ms  (~{suma/1000:.2f} s)")
        print(f"   → Promedio por frame:    {promedio:.2f} ms  (≈{fps_eff:.2f} FPS)\n")

    except Exception as e:
        print(f"Error leyendo {archivo}: {e}")
```
