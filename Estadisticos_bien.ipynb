# Obteniedo el Rango de Frame para cada POSICIÓN y DISTANCIA

```python
import pandas as pd
import matplotlib.pyplot as plt

# ====== RUTAS DE TUS CSVs ======
SEGMENTS_CSV = "segmentos_frames.csv"   # columnas: name,start_frame,end_frame (base 1)
COORDS_CSV   = "capturas_conf_90.csv"   # columnas: frame, *_x, *_y, *_conf

# ====== KEYPOINTS A INCLUIR (tus 4 puntos) ======
KEYPOINTS = [
    "right_shoulder",
    "left_shoulder",
    "right_hip",
    "left_hip",
]

# ====== CARGA Y LIMPIEZA ======
seg = pd.read_csv(SEGMENTS_CSV)
coords = pd.read_csv(COORDS_CSV)

# Tipos y limpieza
for col in ["start_frame", "end_frame"]:
    seg[col] = pd.to_numeric(seg[col], errors="coerce")
coords["frame"] = pd.to_numeric(coords["frame"], errors="coerce")

seg = seg.dropna(subset=["name", "start_frame", "end_frame"]).copy()
coords = coords.dropna(subset=["frame"]).copy()

# Normaliza a enteros
seg["start_frame"] = seg["start_frame"].astype(int)
seg["end_frame"]   = seg["end_frame"].astype(int)
coords["frame"]    = coords["frame"].astype(int)

# Asegura rango correcto (start <= end)
seg["start_frame"], seg["end_frame"] = (
    seg[["start_frame", "end_frame"]].min(axis=1),
    seg[["start_frame", "end_frame"]].max(axis=1),
)

# Verifica que existan columnas necesarias
needed_cols = ["frame"]
for kp in KEYPOINTS:
    needed_cols += [f"{kp}_x", f"{kp}_y", f"{kp}_conf"]
missing = [c for c in needed_cols if c not in coords.columns]
if missing:
    raise ValueError(f"Faltan columnas en {COORDS_CSV}: {missing}")

# ====== GRAFICAR POR SEGMENTO ======
for _, row in seg.iterrows():
    name    = str(row["name"])
    start_f = int(row["start_frame"])
    end_f   = int(row["end_frame"])

    # Subconjunto del rango
    sub = coords[(coords["frame"] >= start_f) & (coords["frame"] <= end_f)].copy()
    if sub.empty:
        print(f"⚠️ Segmento '{name}' ({start_f}-{end_f}) no tiene datos en {COORDS_CSV}.")
        continue

    # Crear figura: 4 filas (keypoints) x 3 columnas (X, Y, Conf)
    fig, axes = plt.subplots(nrows=len(KEYPOINTS), ncols=3, figsize=(14, 10), squeeze=False)
    fig.suptitle(f"{name} — Frames {start_f}-{end_f}", fontsize=14, y=0.995)

    for r, kp in enumerate(KEYPOINTS):
        x_col    = f"{kp}_x"
        y_col    = f"{kp}_y"
        conf_col = f"{kp}_conf"

        # Datos
        frames = sub["frame"].values
        X = sub[x_col].values
        Y = sub[y_col].values
        C = sub[conf_col].values

        # Columna 1: Frame vs X (color = confianza)
        ax1 = axes[r, 0]
        sc1 = ax1.scatter(frames, X, c=C, s=16, cmap="viridis")
        if r == 0:
            ax1.set_title("Frame vs X (color = conf)", fontsize=11)
        ax1.set_ylabel(kp.replace("_", " ").title())
        ax1.set_xlabel("Frame")
        cbar1 = fig.colorbar(sc1, ax=ax1)
        cbar1.set_label("Confianza")

        # Columna 2: Frame vs Y (color = confianza)
        ax2 = axes[r, 1]
        sc2 = ax2.scatter(frames, Y, c=C, s=16, cmap="viridis")
        if r == 0:
            ax2.set_title("Frame vs Y (color = conf)", fontsize=11)
        ax2.set_xlabel("Frame")
        cbar2 = fig.colorbar(sc2, ax=ax2)
        cbar2.set_label("Confianza")
        # NOTA: aquí NO invertimos eje Y porque es Y vs Frame (no coordenadas de imagen)

        # Columna 3: Frame vs Confianza (línea)
        ax3 = axes[r, 2]
        ax3.plot(frames, C, linewidth=1)
        if r == 0:
            ax3.set_title("Frame vs Confianza", fontsize=11)
        ax3.set_xlabel("Frame")
        ax3.set_ylim(0, 1)

    plt.tight_layout(rect=[0, 0, 1, 0.97])
    plt.show()
```

# Variaciones

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# ====== RUTAS DE TUS CSVs ======
SEGMENTS_CSV = "segmentos_frames.csv"   # columnas: name,start_frame,end_frame (base 1)
COORDS_CSV   = "capturas_conf_90.csv"   # columnas: frame, *_x, *_y, *_conf

# ====== CARGA Y LIMPIEZA ======
seg = pd.read_csv(SEGMENTS_CSV)
coords = pd.read_csv(COORDS_CSV)

# Tipos y limpieza
for col in ["start_frame", "end_frame"]:
    seg[col] = pd.to_numeric(seg[col], errors="coerce")
coords["frame"] = pd.to_numeric(coords["frame"], errors="coerce")

seg = seg.dropna(subset=["name", "start_frame", "end_frame"]).copy()
coords = coords.dropna(subset=["frame"]).copy()

# Normaliza a enteros
seg["start_frame"] = seg["start_frame"].astype(int)
seg["end_frame"]   = seg["end_frame"].astype(int)
coords["frame"]    = coords["frame"].astype(int)

# Asegura rango correcto (start <= end)
seg["start_frame"], seg["end_frame"] = (
    seg[["start_frame", "end_frame"]].min(axis=1),
    seg[["start_frame", "end_frame"]].max(axis=1),
)

# Verifica que existan columnas necesarias
needed_cols = [
    "frame",
    "left_shoulder_x", "left_shoulder_y", "left_shoulder_conf",
    "left_hip_x",      "left_hip_y",      "left_hip_conf",
    "right_shoulder_x","right_shoulder_y","right_shoulder_conf",
    "right_hip_x",     "right_hip_y",     "right_hip_conf",
]
missing = [c for c in needed_cols if c not in coords.columns]
if missing:
    raise ValueError(f"Faltan columnas en {COORDS_CSV}: {missing}")

# ====== FUNCIÓN: elige una sola fila por frame (si hay varias) según confianza media relevante ======
def pick_best_per_frame(df):
    """
    Devuelve un DataFrame con como máximo 1 fila por frame.
    Para la distancia izquierda usa avg(left_shoulder_conf, left_hip_conf).
    Para la derecha usa avg(right_shoulder_conf, right_hip_conf).
    Para elegir una sola por frame usamos la media de las dos medias (izq y der).
    """
    df = df.copy()
    left_avg  = (df["left_shoulder_conf"] + df["left_hip_conf"]) / 2.0
    right_avg = (df["right_shoulder_conf"] + df["right_hip_conf"]) / 2.0
    df["_score"] = (left_avg + right_avg) / 2.0

    # Si hay múltiples filas por frame, nos quedamos con la de mayor _score
    df_sorted = df.sort_values(["frame", "_score"], ascending=[True, False])
    df_best = df_sorted.drop_duplicates(subset=["frame"], keep="first").sort_values("frame")
    return df_best.drop(columns=["_score"])

# ====== GRAFICAR POR SEGMENTO ======
for _, row in seg.iterrows():
    name    = str(row["name"])
    start_f = int(row["start_frame"])
    end_f   = int(row["end_frame"])

    sub = coords[(coords["frame"] >= start_f) & (coords["frame"] <= end_f)].copy()
    if sub.empty:
        print(f"⚠️ Segmento '{name}' ({start_f}-{end_f}) no tiene datos en {COORDS_CSV}.")
        continue

    # Elegir una sola detección por frame si hubiese varias
    sub = pick_best_per_frame(sub)

    # Elimina filas con NaN en cualquiera de las columnas necesarias para las distancias
    req = [
        "left_shoulder_x","left_shoulder_y","left_hip_x","left_hip_y",
        "right_shoulder_x","right_shoulder_y","right_hip_x","right_hip_y"
    ]
    sub = sub.dropna(subset=req)

    # Distancias euclidianas
    dist_left = np.sqrt(
        (sub["left_shoulder_x"] - sub["left_hip_x"])**2 +
        (sub["left_shoulder_y"] - sub["left_hip_y"])**2
    )
    dist_right = np.sqrt(
        (sub["right_shoulder_x"] - sub["right_hip_x"])**2 +
        (sub["right_shoulder_y"] - sub["right_hip_y"])**2
    )

    frames = sub["frame"].values

    # Figura con 2 subplots: izquierda y derecha
    fig, axes = plt.subplots(nrows=2, ncols=1, figsize=(10, 6), sharex=True)
    fig.suptitle(f"{name} — Frames {start_f}-{end_f} — Distancia euclidiana hombro–cadera", fontsize=14, y=0.97)

    # Izquierda
    axes[0].plot(frames, dist_left, linewidth=1.2)
    axes[0].set_ylabel("Distancia Izquierda (px)")
    axes[0].grid(True, alpha=0.25)

    # Derecha
    axes[1].plot(frames, dist_right, linewidth=1.2)
    axes[1].set_xlabel("Frame")
    axes[1].set_ylabel("Distancia Derecha (px)")
    axes[1].grid(True, alpha=0.25)

    plt.tight_layout(rect=[0, 0, 1, 0.95])
    plt.show()
```

# variación por segmentos

```python
SEGMENTS_CSV = "segmentos_frames.csv"     # columnas: name,start_frame,end_frame (base 1)
COORDS_CSV   = "capturas_conf_90.csv"     # columnas: frame, *_x, *_y, *_conf

# ====== CARGA ======
seg = pd.read_csv(SEGMENTS_CSV)
coords = pd.read_csv(COORDS_CSV)

# Tipos y limpieza
for col in ["start_frame", "end_frame"]:
    seg[col] = pd.to_numeric(seg[col], errors="coerce")
coords["frame"] = pd.to_numeric(coords["frame"], errors="coerce")

seg = seg.dropna(subset=["name", "start_frame", "end_frame"]).copy()
coords = coords.dropna(subset=["frame"]).copy()

seg["start_frame"] = seg["start_frame"].astype(int)
seg["end_frame"]   = seg["end_frame"].astype(int)
coords["frame"]    = coords["frame"].astype(int)

# Asegura rangos válidos
seg["start_frame"], seg["end_frame"] = (
    seg[["start_frame", "end_frame"]].min(axis=1),
    seg[["start_frame", "end_frame"]].max(axis=1),
)

# Columnas requeridas
needed = [
    "frame",
    "left_shoulder_x","left_shoulder_y","left_shoulder_conf",
    "left_hip_x","left_hip_y","left_hip_conf",
    "right_shoulder_x","right_shoulder_y","right_shoulder_conf",
    "right_hip_x","right_hip_y","right_hip_conf"
]
missing = [c for c in needed if c not in coords.columns]
if missing:
    raise ValueError(f"Faltan columnas en {COORDS_CSV}: {missing}")

def pick_best_per_frame(df):
    """
    Si hay varias filas por frame, escoge la mejor por confianza.
    Score = promedio de (promedio conf hombro izq + cadera izq, promedio conf hombro der + cadera der).
    """
    df = df.copy()
    left_avg  = (df["left_shoulder_conf"] + df["left_hip_conf"]) / 2.0
    right_avg = (df["right_shoulder_conf"] + df["right_hip_conf"]) / 2.0
    df["_score"] = (left_avg + right_avg) / 2.0
    df = df.sort_values(["frame", "_score"], ascending=[True, False])
    df = df.drop_duplicates(subset=["frame"], keep="first").sort_values("frame")
    return df.drop(columns=["_score"])

def euclid(ax, ay, bx, by):
    return np.sqrt((ax - bx)**2 + (ay - by)**2)

# ====== POR SEGMENTO: distancia y variación (|Δ|) ======
for _, row in seg.iterrows():
    name    = str(row["name"])
    start_f = int(row["start_frame"])
    end_f   = int(row["end_frame"])

    sub = coords[(coords["frame"] >= start_f) & (coords["frame"] <= end_f)].copy()
    if sub.empty:
        print(f"⚠️ Segmento '{name}' ({start_f}-{end_f}) no tiene filas en {COORDS_CSV}.")
        continue

    # Una detección por frame (si hubiera varias)
    sub = pick_best_per_frame(sub)

    # Asegura datos completos para distancias
    req = [
        "left_shoulder_x","left_shoulder_y","left_hip_x","left_hip_y",
        "right_shoulder_x","right_shoulder_y","right_hip_x","right_hip_y"
    ]
    sub = sub.dropna(subset=req).copy()

    # Distancias (px)
    dist_left  = euclid(sub["left_shoulder_x"],  sub["left_shoulder_y"],
                        sub["left_hip_x"],       sub["left_hip_y"])
    dist_right = euclid(sub["right_shoulder_x"], sub["right_shoulder_y"],
                        sub["right_hip_x"],      sub["right_hip_y"])

    frames = sub["frame"].to_numpy()

    # Variación frame a frame: |Δ distancia|
    var_left  = np.abs(np.diff(dist_left, prepend=dist_left.iloc[0]))
    var_right = np.abs(np.diff(dist_right, prepend=dist_right.iloc[0]))

    # ====== FIGURA 2x2 ======
    fig, axes = plt.subplots(2, 2, figsize=(12, 7), sharex=True)
    fig.suptitle(f"{name} — Frames {start_f}-{end_f} — Distancia y variación hombro–cadera", y=0.98)

    # Izquierda: distancia
    axes[0, 0].plot(frames, dist_left, linewidth=1.2)
    axes[0, 0].set_ylabel("Dist. Izquierda (px)")
    axes[0, 0].grid(True, alpha=0.25)
    axes[0, 0].set_title("Distancia izquierda")

    # Izquierda: variación
    axes[1, 0].plot(frames, var_left, linewidth=1.2)
    axes[1, 0].set_xlabel("Frame")
    axes[1, 0].set_ylabel("|Δ| Izquierda (px)")
    axes[1, 0].grid(True, alpha=0.25)
    axes[1, 0].set_title("Variación |Δ| izquierda")

    # Derecha: distancia
    axes[0, 1].plot(frames, dist_right, linewidth=1.2)
    axes[0, 1].set_ylabel("Dist. Derecha (px)")
    axes[0, 1].grid(True, alpha=0.25)
    axes[0, 1].set_title("Distancia derecha")

    # Derecha: variación
    axes[1, 1].plot(frames, var_right, linewidth=1.2)
    axes[1, 1].set_xlabel("Frame")
    axes[1, 1].set_ylabel("|Δ| Derecha (px)")
    axes[1, 1].grid(True, alpha=0.25)
    axes[1, 1].set_title("Variación |Δ| derecha")

    plt.tight_layout(rect=[0, 0, 1, 0.95])
    plt.show()
```

```python
SEGMENTS_CSV = "segmentos_frames.csv"     # columnas: name,start_frame,end_frame
COORDS_CSV   = "capturas_conf_90.csv"     # columnas: frame, *_x, *_y, *_conf

seg = pd.read_csv(SEGMENTS_CSV)
coords = pd.read_csv(COORDS_CSV)

# Tipos y limpieza
for col in ["start_frame", "end_frame"]:
    seg[col] = pd.to_numeric(seg[col], errors="coerce")
coords["frame"] = pd.to_numeric(coords["frame"], errors="coerce")

seg = seg.dropna(subset=["name", "start_frame", "end_frame"]).copy()
coords = coords.dropna(subset=["frame"]).copy()

seg["start_frame"] = seg["start_frame"].astype(int)
seg["end_frame"]   = seg["end_frame"].astype(int)
coords["frame"]    = coords["frame"].astype(int)

# Asegura rangos válidos
seg["start_frame"], seg["end_frame"] = (
    seg[["start_frame", "end_frame"]].min(axis=1),
    seg[["start_frame", "end_frame"]].max(axis=1),
)

# Columnas necesarias
needed = [
    "frame",
    "left_shoulder_x","left_shoulder_y","left_hip_x","left_hip_y",
    "right_shoulder_x","right_shoulder_y","right_hip_x","right_hip_y"
]
missing = [c for c in needed if c not in coords.columns]
if missing:
    raise ValueError(f"Faltan columnas en {COORDS_CSV}: {missing}")

def euclid(ax, ay, bx, by):
    return np.sqrt((ax - bx)**2 + (ay - by)**2)

# ====== PROCESAR POR SEGMENTO ======
for _, row in seg.iterrows():
    name    = str(row["name"])
    start_f = int(row["start_frame"])
    end_f   = int(row["end_frame"])

    sub = coords[(coords["frame"] >= start_f) & (coords["frame"] <= end_f)].copy()
    if sub.empty:
        print(f"⚠️ Segmento '{name}' ({start_f}-{end_f}) no tiene filas en {COORDS_CSV}.")
        continue

    # Elimina filas con NaN en las columnas críticas
    sub = sub.dropna(subset=needed)

    # Distancias
    dist_left  = euclid(sub["left_shoulder_x"],  sub["left_shoulder_y"],
                        sub["left_hip_x"],       sub["left_hip_y"])
    dist_right = euclid(sub["right_shoulder_x"], sub["right_shoulder_y"],
                        sub["right_hip_x"],      sub["right_hip_y"])

    # Promedio
    dist_avg = (dist_left + dist_right) / 2.0

    # Estadísticos
    mean_val = np.mean(dist_avg)
    var_val  = np.var(dist_avg, ddof=0)  # varianza poblacional
    std_val  = np.std(dist_avg, ddof=0)

    print(f"📊 Segmento '{name}' ({start_f}-{end_f})")
    print(f"   Media: {mean_val:.2f} px")
    print(f"   Varianza: {var_val:.2f}")
    print(f"   Desv. estándar: {std_val:.2f}")

    frames = sub["frame"].to_numpy()

    # ====== GRAFICAR ======
    plt.figure(figsize=(10, 5))
    plt.plot(frames, dist_avg, label="Distancia promedio hombro–cadera", linewidth=1.2)

    # Media
    plt.axhline(mean_val, color="red", linestyle="--", linewidth=1.2, label=f"Media = {mean_val:.2f}")
    # Banda ±desviación
    plt.axhline(mean_val + std_val, color="orange", linestyle=":", linewidth=1.2, label=f"+1σ = {mean_val+std_val:.2f}")
    plt.axhline(mean_val - std_val, color="orange", linestyle=":", linewidth=1.2, label=f"-1σ = {mean_val-std_val:.2f}")

    plt.xlabel("Frame")
    plt.ylabel("Distancia promedio (px)")
    plt.title(f"{name} | Frames {start_f}-{end_f}\nMedia={mean_val:.2f}, Var={var_val:.2f}, σ={std_val:.2f}")
    plt.legend()
    plt.grid(alpha=0.3)
    plt.tight_layout()
    plt.show()
```
