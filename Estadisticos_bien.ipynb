Confianza de los puntos por frame del hombro derecho

```python
import pandas as pd
import matplotlib.pyplot as plt

# Carga el CSV
df = pd.read_csv("capturas_conf_90.csv")

# Si a veces hay varias personas por frame y quieres solo una,
# puedes quedarte con la detección de mayor confianza de HOMBRO DERECHO por frame:
df = (df
      .sort_values(["frame", "right_shoulder_conf"], ascending=[True, False])
      .drop_duplicates(subset=["frame"])
      .sort_values("frame"))

# Extrae series
frames = df["frame"].values
x_rs = df["right_shoulder_x"].values
y_rs = df["right_shoulder_y"].values
c_rs = df["right_shoulder_conf"].values

plt.figure(figsize=(12, 9))

# 1) Frame vs X (color por confianza)
plt.subplot(3, 1, 1)
sc1 = plt.scatter(frames, x_rs, c=c_rs, cmap="viridis", s=20)
plt.ylabel("Right Shoulder X (px)")
plt.title("Right Shoulder: Frame vs X (color = confianza)")
cbar1 = plt.colorbar(sc1)
cbar1.set_label("Confianza")

# 2) Frame vs Y (color por confianza)
plt.subplot(3, 1, 2)
sc2 = plt.scatter(frames, y_rs, c=c_rs, cmap="viridis", s=20)
plt.ylabel("Right Shoulder Y (px)")
plt.title("Right Shoulder: Frame vs Y (color = confianza)")
cbar2 = plt.colorbar(sc2)
cbar2.set_label("Confianza")
# Nota: no invertimos eje Y aquí porque ahora el eje Y representa valor vs frame (no coordenadas en imagen)

# 3) Frame vs Confianza (línea)
plt.subplot(3, 1, 3)
plt.plot(frames, c_rs, linewidth=1)
plt.xlabel("Frame")
plt.ylabel("Confianza")
plt.title("Right Shoulder: Confianza vs Frame")

plt.tight_layout()
plt.show()
```

# Obteniedo el Rango de Frame para cada POSICIÓN y DISTANCIA

```python
import pandas as pd
import matplotlib.pyplot as plt

# ====== RUTAS DE TUS CSVs ======
SEGMENTS_CSV = "segmentos_frames.csv"   # columnas: name,start_frame,end_frame (base 1)
COORDS_CSV   = "capturas_conf_90.csv"   # columnas: frame, *_x, *_y, *_conf

# ====== KEYPOINTS A INCLUIR (tus 4 puntos) ======
KEYPOINTS = [
    "right_shoulder",
    "left_shoulder",
    "right_hip",
    "left_hip",
]

# ====== CARGA Y LIMPIEZA ======
seg = pd.read_csv(SEGMENTS_CSV)
coords = pd.read_csv(COORDS_CSV)

# Tipos y limpieza
for col in ["start_frame", "end_frame"]:
    seg[col] = pd.to_numeric(seg[col], errors="coerce")
coords["frame"] = pd.to_numeric(coords["frame"], errors="coerce")

seg = seg.dropna(subset=["name", "start_frame", "end_frame"]).copy()
coords = coords.dropna(subset=["frame"]).copy()

# Normaliza a enteros
seg["start_frame"] = seg["start_frame"].astype(int)
seg["end_frame"]   = seg["end_frame"].astype(int)
coords["frame"]    = coords["frame"].astype(int)

# Asegura rango correcto (start <= end)
seg["start_frame"], seg["end_frame"] = (
    seg[["start_frame", "end_frame"]].min(axis=1),
    seg[["start_frame", "end_frame"]].max(axis=1),
)

# Verifica que existan columnas necesarias
needed_cols = ["frame"]
for kp in KEYPOINTS:
    needed_cols += [f"{kp}_x", f"{kp}_y", f"{kp}_conf"]
missing = [c for c in needed_cols if c not in coords.columns]
if missing:
    raise ValueError(f"Faltan columnas en {COORDS_CSV}: {missing}")

# ====== GRAFICAR POR SEGMENTO ======
for _, row in seg.iterrows():
    name    = str(row["name"])
    start_f = int(row["start_frame"])
    end_f   = int(row["end_frame"])

    # Subconjunto del rango
    sub = coords[(coords["frame"] >= start_f) & (coords["frame"] <= end_f)].copy()
    if sub.empty:
        print(f"⚠️ Segmento '{name}' ({start_f}-{end_f}) no tiene datos en {COORDS_CSV}.")
        continue

    # Crear figura: 4 filas (keypoints) x 3 columnas (X, Y, Conf)
    fig, axes = plt.subplots(nrows=len(KEYPOINTS), ncols=3, figsize=(14, 10), squeeze=False)
    fig.suptitle(f"{name} — Frames {start_f}-{end_f}", fontsize=14, y=0.995)

    for r, kp in enumerate(KEYPOINTS):
        x_col    = f"{kp}_x"
        y_col    = f"{kp}_y"
        conf_col = f"{kp}_conf"

        # Datos
        frames = sub["frame"].values
        X = sub[x_col].values
        Y = sub[y_col].values
        C = sub[conf_col].values

        # Columna 1: Frame vs X (color = confianza)
        ax1 = axes[r, 0]
        sc1 = ax1.scatter(frames, X, c=C, s=16, cmap="viridis")
        if r == 0:
            ax1.set_title("Frame vs X (color = conf)", fontsize=11)
        ax1.set_ylabel(kp.replace("_", " ").title())
        ax1.set_xlabel("Frame")
        cbar1 = fig.colorbar(sc1, ax=ax1)
        cbar1.set_label("Confianza")

        # Columna 2: Frame vs Y (color = confianza)
        ax2 = axes[r, 1]
        sc2 = ax2.scatter(frames, Y, c=C, s=16, cmap="viridis")
        if r == 0:
            ax2.set_title("Frame vs Y (color = conf)", fontsize=11)
        ax2.set_xlabel("Frame")
        cbar2 = fig.colorbar(sc2, ax=ax2)
        cbar2.set_label("Confianza")
        # NOTA: aquí NO invertimos eje Y porque es Y vs Frame (no coordenadas de imagen)

        # Columna 3: Frame vs Confianza (línea)
        ax3 = axes[r, 2]
        ax3.plot(frames, C, linewidth=1)
        if r == 0:
            ax3.set_title("Frame vs Confianza", fontsize=11)
        ax3.set_xlabel("Frame")
        ax3.set_ylim(0, 1)

    plt.tight_layout(rect=[0, 0, 1, 0.97])
    plt.show()
```
